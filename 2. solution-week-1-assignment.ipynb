### Design a multi-line plot using Matplotlib or Seaborn that shows the evolution of average mpg over the years, separately for each origin

```python
import pandas as pd

cars_df = pd.read_csv('/kaggle/input/cars-csv/Cars.csv')

# Print the column names to see what you're working with
print(cars_df.columns)

# Display the first few rows to get a sense of the data
print(cars_df.head())
```

```python
# Importing libraries
import warnings
warnings.filterwarnings('ignore')

import matplotlib.pyplot as plt
import seaborn as sns

cars_df_reset = cars_df.reset_index()

avg_mpg = cars_df_reset.groupby(['model_year', 'origin'])['mpg'].mean().reset_index()

sns.set(style="whitegrid")

plt.figure(figsize=(12, 7))
sns.lineplot(
    data=avg_mpg,
    x='model_year',
    y='mpg',
    hue='origin',
    marker='o',
    palette='tab10'
)


plt.title('Average MPG Over the Years by Origin', fontsize=16)
plt.xlabel('Model Year', fontsize=12)
plt.ylabel('Average MPG', fontsize=12)
plt.legend(title='Origin', title_fontsize=12)
plt.xticks(avg_mpg['model_year'].unique())
plt.tight_layout()
plt.show()
```

### We define a “consistent” car model as one that was produced over multiple years and had very low variation in mpg across those years (standard deviation < 1.0).

Tasks:

Identify car names that appear in more than one model_year.

For each such name, compute the standard deviation of mpg across years.

Return the car(s) with the lowest variation in mpg, among those with at least 2 appearances and std(mpg) < 1.0.

Report the model name(s), number of appearances, and the average mpg.

Bonus:
Sort the result by number of appearances (descending), then mpg (descending).

```python
import pandas as pd

#Identify car names that appear in more than one model_year
cars_df['base_name'] = cars_df['name'].str.replace(r'\([^)]*\)', '', regex=True)
cars_df['base_name'] = cars_df['base_name'].str.replace(r'\d{4}', '', regex=True)
cars_df['base_name'] = cars_df['base_name'].str.strip()  # Trim whitespace
model_counts = cars_df.groupby('base_name')['model_year'].nunique()
multi_year_models = model_counts[model_counts > 1].index.tolist()

#For each multi-year model, compute std of mpg
results = []
for model in multi_year_models:
    model_data = cars_df[cars_df['base_name'] == model]
    mpg_std = model_data['mpg'].std()
    mpg_mean = model_data['mpg'].mean()
    count = len(model_data)

    if mpg_std < 1.0 and count >= 2:
        results.append({
            'model': model,
            'appearances': count,
            'avg_mpg': round(mpg_mean, 2),
            'mpg_std': round(mpg_std, 4)
        })

#Find the model(s) with lowest variation (smallest std)
if results:
    results_df = pd.DataFrame(results)
    min_std = results_df['mpg_std'].min()
    most_consistent = results_df[results_df['mpg_std'] == min_std]
    most_consistent = most_consistent.sort_values(['appearances', 'avg_mpg'], ascending=[False, False])
    print("Most consistent car model(s) with lowest MPG variation:")
    print(most_consistent.to_string(index=False))
else:
    print("No models found with multiple years and MPG standard deviation < 1.0")
```
